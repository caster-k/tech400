## Creating 5 Documents

```python
import re

documents = {
    "doc1": """
    Artificial intelligence allows computers to perform tasks that normally
    require human intelligence. AI is used in learning, reasoning, and decision making.
    """,

    "doc2": """
    Machine learning is a branch of artificial intelligence. Machine learning
    allows computers to learn patterns from data and make predictions.
    """,

    "doc3": """
    Deep learning uses neural networks to learn complex patterns from large datasets.
    Deep learning is widely used in computer vision and natural language processing.
    """,

    "doc4": """
    Natural language processing allows computers to understand and process human language.
    NLP is used in translation, chatbots, and text classification.
    """,

    "doc5": """
    Computer vision enables computers to understand images and videos.
    Robotics combines computer vision, artificial intelligence, and machine learning
    to create intelligent robots.
    """
}
```

## Adding Documents to the system

```python
def add_document(doc_id, text):
    documents[doc_id] = text
add_document(
    "doc6",
    "Robotics combines artificial intelligence and machine learning."
)

documents
```

## Cleaning, stemming and removing the stop words

```python

# 1. Clean text
def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    return text.split()


# 2. Remove stop words
def remove_stop_words(words):
    stop_words = {
        "a", "an", "the", "to", "and",
        "is", "of", "in", "on", "for"
    }

    return [word for word in words if word not in stop_words]


# 3. Convert words to simple root/stem words
def stem_word(word):

    if word.endswith("ing"):
        return word[:-3]

    if word.endswith("tion"):
        return word[:-3]

    if word.endswith("ics"):
        return word[:-3]

    if word.endswith("s") and len(word) > 3:
        return word[:-1]

    return word


def stem_words(words):
    return [stem_word(word) for word in words]


processed_documents = {}

for doc_id, text in documents.items():
    words = clean_text(text)
    words = remove_stop_words(words)
    processed_documents[doc_id] = stem_words(words)
```

## Making a basic dictionary

```python
def create_dictionary(documents):
    dictionary = set()

    for words in documents.values():
        dictionary.update(words)

    return dictionary

dictionary = create_dictionary(processed_documents)

print("Dictionary:")
print(sorted(dictionary))
```

## Inverted Index

```python
def create_inverted_index(processed_documents):

    inverted_index = {}

    for doc_id, words in processed_documents.items():

        for word in words:

            if word not in inverted_index:
                inverted_index[word] = set()

            inverted_index[word].add(doc_id)

    return inverted_index

inverted_index = create_inverted_index(processed_documents)

for word in sorted(inverted_index):
    print(word, ":", sorted(inverted_index[word]))
```

## Adding Boolean Retrieval

### AND

```python
def boolean_and(term1, term2):
    return inverted_index.get(term1, set()) & inverted_index.get(term2, set())
print(boolean_and("machine", "learning"))
```

### OR

```python
def boolean_or(term1, term2):
    return inverted_index.get(term1, set()) | inverted_index.get(term2, set())
print(boolean_or("computer", "robotics"))
```

### NOT

```python
def boolean_not(term):
    all_documents = set(documents.keys())

    return all_documents - inverted_index.get(term, set())
print(boolean_not("robotics"))
```
